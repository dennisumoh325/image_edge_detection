```python
import cv2
import numpy as np
import matplotlib.pyplot as plt
%matplotlib inline
```

### The Prewitt filter

```python
car = cv2.imread(r"C:\Users\New\Desktop\Jupyter Notebook\ML\Computer Vision\edge_detection_CV\640px-1995Ferrari456GT.jpg")

cv2.imshow('Image', car)
plt.imshow(cv2.cvtColor(car, cv2.COLOR_BGR2RGB))
```

```python
kernel_x= np.array([[-1, -1, -1],
                   [0,  0, 0],
                   [1, 1, 1]])

kernel_y =np.array([[-1, 0, 1],
                   [-1,  0, 1],
                   [-1, 0, 1]])

print("Horizontal Edge detector:")
print(kernel_x)
print("Vertical Edge detector:")
print(kernel_y)
```

```python
perwitt_x = cv2.filter2D(car, -1, kernel_x)
perwitt_y = cv2.filter2D(car, -1, kernel_y)

perwitt_xy = cv2.bitwise_or(perwitt_x, perwitt_y)
```

```python
frame_prewitt =cv2.hconcat([perwitt_x, perwitt_y, perwitt_xy])

cv2.imshow('Prewitt Edge Detection', frame_prewitt)
plt.imshow(cv2.cvtColor(frame_prewitt, cv2.COLOR_BGR2RGB))
```
